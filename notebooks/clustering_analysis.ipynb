```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Load your downloaded data from the data folder
# (Replace with your actual filename if it matches exactly)
df = pd.read_csv('../data/all_tickets_processed_improved_v3.csv')

# Let's inspect the column names to make sure we match your data preview
print("Available columns:", df.columns)

# Assume 'Document' is our text column and 'Topic_group' is our category label
text_col = 'Document' 

# 2. Initialize the TF-IDF Vectorizer
# max_features=2500 keeps the top 2500 most important words
# stop_words='english' filters out standard English noise words
tfidf = TfidfVectorizer(max_features=2500, stop_words='english')

# 3. Fit and transform the text column into a mathematical matrix
print("Vectorizing text data... this may take a moment.")
X = tfidf.fit_transform(df[text_col].astype(str))

print(f"Text successfully converted to matrix shape: {X.shape}")
print(f"That means we have {X.shape[0]} tickets mapped across {X.shape[1]} unique word features!")
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# To save time on 47k rows, we calculate the elbow curve using a mini-batch K-Means 
# or a sample of the data. Let's use a sample of 10,000 rows for the graph calculation.
X_sample = X[:10000]

inertia_scores = []
k_range = range(2, 11)

print("Calculating elbow scores for K=2 through K=10...")
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=5)
    kmeans.fit(X_sample)
    inertia_scores.append(kmeans.inertia_)
    print(f"Finished K={k}")

# Plotting the Elbow Curve
plt.figure(figsize=(8, 5))
plt.plot(k_range, inertia_scores, marker='o', linestyle='--', color='b')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia (Within-Cluster Sum of Squares)')
plt.title('The Elbow Method for Optimal K')
plt.xticks(k_range)
plt.grid(True)
plt.show()
```
